```python
def OddEven():
    num=int(input('Enter a number:'))
    if(num%2==0):
        print(f'{num}is even number')
    else:
        print('No is odd number')
OddEven()      
```

```python
def Eligible():
    Gender=input('Your Gender:')
    Age=int(input('Your Age'))
    if Gender=='male':
        if(Age>=21):
            print('ELIGIBLE')
        else:
            print('NOT ELIGIBLE')
    elif Gender=='female':
        if(Age>=18):
            print('ELIGIBLE')
        else:
            print('NOT ELIGIBLE')
    else:
        print('Invalid Gender')
Eligible()        
    
```
