```python
def user_vector(ratings,movie_vector):
    vector=[]
    for user in range(1,944):
        u1=ratings[ratings['user_id']==user][['movie_id','rating']]
        u1_list=[]
        for i in u1['movie_id']:
            f=movie_vector[i].values
            u1_list.append(f)
            for j in range(1,19):
                g=sum(uu1[j])
                vec.append(g/len(u1))
                vector.append(vec)
        return vector
     
    
```
