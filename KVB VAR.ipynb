```python
ohlc = data[['Open', 'High', 'Low', 'Close']].copy()

if hasattr(ohlc.columns, 'levels'):
    ohlc.columns = ohlc.columns.get_level_values(0)

ohlc = ohlc.dropna()

print(ohlc.head())
```

```python
ohlc_diff = ohlc.diff().dropna()
```

```python
train_size = int(len(ohlc_diff) * 0.8)

train = ohlc_diff.iloc[:train_size]
test = ohlc_diff.iloc[train_size:]

print("Training observations:", len(train))
print("Testing observations :", len(test))
```

```python
from statsmodels.tsa.api import VAR

model = VAR(train)

var_model = model.fit(3)

print("VAR(3) model fitted successfully")
```

```python
import pandas as pd
forecast = var_model.forecast(
    train.values[-3:],
    steps=len(test)
)

forecast_df = pd.DataFrame(
    forecast,
    index=test.index,
    columns=test.columns
)

print(forecast_df)
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

for column in test.columns:
    rmse = np.sqrt(
        mean_squared_error(
            test[column],
            forecast_df[column]
        )
    )

    print(column, "RMSE :", rmse)
```

```python
# Forecast next trading day
next_diff = var_model.forecast(
    ohlc_diff.values[-3:],
    steps=1
)

next_diff = pd.DataFrame(
    next_diff,
    columns=['Open', 'High', 'Low', 'Close']
)

print("Predicted Price Change:")
print(next_diff)
```

```python
# Convert predicted difference to actual OHLC price

last_ohlc = ohlc.iloc[-1]

next_ohlc = last_ohlc + next_diff.iloc[0]

print("\nPredicted Next Trading Day OHLC:")
print(next_ohlc)
```
